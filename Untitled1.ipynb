```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import optimize
from sklearn.cluster import KMeans
from sklearn.svm import SVC
from sklearn.datasets import load_iris, make_blobs, make_circles, load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

rng = np.random.default_rng(422)
plt.rcParams["figure.figsize"] = (7, 4)
plt.rcParams["axes.grid"] = True
```

#3.5 K-Means

Goal: split n points x₁, …, xₙ ∈ Rᵈ into k clusters C₁, …, Cₖ with centers μ₁, …, μₖ that minimize the within-cluster sum of squares: G(C, μ) = ∑ⱼ ∑_{i ∈ Cⱼ} ‖xᵢ − μⱼ‖²

Finding the global minimum is NP-hard, so we alternate between the two variables (Lloyd's algorithm). Step 1, assignment (μ fixed): put each point in the cluster of its nearest center, c(i) = argminⱼ ‖xᵢ − μⱼ‖²

Step 2, update (C fixed): each center becomes the mean of its cluster, μⱼ = (1/|Cⱼ|) ∑_{i ∈ Cⱼ} xᵢ

Why the mean: ∑ ‖xᵢ − μ‖² is a convex function of μ with gradient 2∑(μ − xᵢ), which is 0 exactly at the mean.

Properties:
- Neither step can increase G, so G decreases monotonically.
- There are finitely many partitions, so the algorithm stops after finitely many steps.
- It stops at a local minimum, and the result depends on the starting centers.

```python
def kmeans(X, k, init, max_iter=100):
    mu = init.copy()
    history = []
    for _ in range(max_iter):
        d2 = ((X[:, None, :] - mu[None, :, :]) ** 2).sum(axis=2)
        labels = d2.argmin(axis=1)
        history.append(d2[np.arange(len(X)), labels].sum())
        new_mu = np.array([X[labels == j].mean(axis=0) if np.any(labels == j) else mu[j] for j in range(k)])
        if np.allclose(new_mu, mu):
            break
        mu = new_mu
    return labels, mu, np.array(history)

def wcss(X, labels, mu):
    return ((X - mu[labels]) ** 2).sum()
```

##Example 1: Lloyd's algorithm step by step

Three clusters of simulated points. The plots show the assignments and centers after iterations 1, 2 and the final one, and the objective G after each assignment step.

```python
X, y_true = make_blobs(n_samples=300, centers=[[0, 0], [6, 5], [6, -3]], cluster_std=1.2, random_state=4)
k = 3
init = np.array([[-3.0, 7.0], [1.0, -6.0], [10.0, 8.0]])
labels, mu, hist = kmeans(X, k, init)

print("G after each iteration:", np.round(hist, 2))
print("monotone decreasing:   ", np.all(np.diff(hist) <= 1e-9))
print("iterations to converge:", len(hist))

fig, ax = plt.subplots(1, 4, figsize=(17, 3.8))
for a, it in zip(ax[:3], [1, 2, len(hist)]):
    lab_it, mu_it, _ = kmeans(X, k, init, max_iter=it)
    a.scatter(X[:, 0], X[:, 1], c=lab_it, cmap="viridis", s=12)
    a.scatter(mu_it[:, 0], mu_it[:, 1], c="red", marker="X", s=200, edgecolor="k")
    a.set_title(f"after {it} iteration(s)")
ax[3].plot(np.arange(1, len(hist) + 1), hist, "o-")
ax[3].set_xlabel("iteration"); ax[3].set_title("objective G")
plt.tight_layout(); plt.show()
```

##Example 2: Comparison with scikit-learn

scikit-learn's KMeans runs the same algorithm. Its `inertia_` is G. Started from the same centers, both should reach the same clusters and the same G.

```python
sk = KMeans(n_clusters=k, init=init, n_init=1).fit(X)
print("our G:    ", round(wcss(X, labels, mu), 4))
print("sklearn G:", round(sk.inertia_, 4))
print("same centers (any order):", np.allclose(np.sort(mu, axis=0), np.sort(sk.cluster_centers_, axis=0)))

g = np.linspace(X.min() - 1, X.max() + 1, 300)
G1, G2 = np.meshgrid(g, g)
grid_lab = ((np.c_[G1.ravel(), G2.ravel()][:, None, :] - mu[None]) ** 2).sum(axis=2).argmin(axis=1)
plt.figure(figsize=(6, 5))
plt.contourf(G1, G2, grid_lab.reshape(G1.shape), alpha=0.2, cmap="viridis")
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis", s=12)
plt.scatter(mu[:, 0], mu[:, 1], c="red", marker="X", s=200, edgecolor="k")
plt.xlim(X[:, 0].min() - 1, X[:, 0].max() + 1); plt.ylim(X[:, 1].min() - 1, X[:, 1].max() + 1)
plt.title("Final clusters; shaded regions = nearest-center (Voronoi) cells")
plt.show()
```

The shaded regions show that each cluster is the set of points closest to its center (a Voronoi cell). The boundaries between cells are straight lines, so K-means works best on compact, roughly round clusters.

##Example 3: Local minima and initialization

Each random start can end at a different local minimum. Two common fixes:
- Run several random starts and keep the lowest G.
- k-means++: choose the first center at random, then choose each next center with probability proportional to its squared distance from the nearest center already chosen. This spreads the starting centers out.

```python
X4, _ = make_blobs(n_samples=400, centers=[[-6, 0], [-2, 0], [2, 0], [6, 0]], cluster_std=0.6, random_state=1)

def kmeanspp_init(X, k, rng):
    centers = [X[rng.integers(len(X))]]
    for _ in range(k - 1):
        d2 = np.min(((X[:, None, :] - np.array(centers)[None]) ** 2).sum(axis=2), axis=1)
        centers.append(X[rng.choice(len(X), p=d2 / d2.sum())])
    return np.array(centers)

rand_G = [wcss(X4, *kmeans(X4, 4, X4[rng.choice(len(X4), 4, replace=False)])[:2]) for _ in range(200)]
pp_G = [wcss(X4, *kmeans(X4, 4, kmeanspp_init(X4, 4, rng))[:2]) for _ in range(200)]
best = min(rand_G + pp_G)
print(f"best G found: {best:.1f}")
print(f"random init reaches best G: {np.mean(np.isclose(rand_G, best)):.0%} of runs")
print(f"k-means++   reaches best G: {np.mean(np.isclose(pp_G, best)):.0%} of runs")

worst = X4[rng.choice(len(X4), 4, replace=False)]
for _ in range(200):
    cand = X4[rng.choice(len(X4), 4, replace=False)]
    if wcss(X4, *kmeans(X4, 4, cand)[:2]) > wcss(X4, *kmeans(X4, 4, worst)[:2]):
        worst = cand
lab_bad, mu_bad, _ = kmeans(X4, 4, worst)
lab_good, mu_good, _ = kmeans(X4, 4, kmeanspp_init(X4, 4, np.random.default_rng(0)))

fig, ax = plt.subplots(1, 3, figsize=(16, 3.8))
ax[0].hist([rand_G, pp_G], bins=20, label=["random init", "k-means++"])
ax[0].set_xlabel("final G"); ax[0].set_title("Final objective over 200 runs"); ax[0].legend()
for a, lab, m, t in [(ax[1], lab_bad, mu_bad, "bad local minimum"), (ax[2], lab_good, mu_good, "best G found")]:
    a.scatter(X4[:, 0], X4[:, 1], c=lab, cmap="viridis", s=10)
    a.scatter(m[:, 0], m[:, 1], c="red", marker="X", s=200, edgecolor="k")
    a.set_title(f"{t}, G = {wcss(X4, lab, m):.0f}")
plt.tight_layout(); plt.show()
```

##Example 4: Choosing k (elbow method)

The minimum G always decreases as k grows (G = 0 when k = n), so we can't pick k by minimizing G. Instead, plot G against k and pick the "elbow", where adding another cluster stops reducing G by much.

Data: the iris dataset (150 flowers, 4 measurements, 3 species). The species labels are not used for clustering. They are only used afterwards to check the clusters.

```python
iris = load_iris()
Xi = StandardScaler().fit_transform(iris.data)
ks = range(1, 9)
G_k = [KMeans(n_clusters=kk, n_init=10, random_state=0).fit(Xi).inertia_ for kk in ks]

km3 = KMeans(n_clusters=3, n_init=10, random_state=0).fit(Xi)
table = np.array([[np.sum((km3.labels_ == c) & (iris.target == s)) for s in range(3)] for c in range(3)])
print("rows = clusters, columns =", iris.target_names.tolist())
print(table)

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(list(ks), G_k, "o-")
ax[0].set_xlabel("k"); ax[0].set_ylabel("G"); ax[0].set_title("Elbow plot (iris, standardized)")
ax[1].scatter(Xi[:, 2], Xi[:, 3], c=km3.labels_, cmap="viridis", s=15)
ax[1].set_xlabel("petal length (std)"); ax[1].set_ylabel("petal width (std)"); ax[1].set_title("K-means clusters, k = 3")
plt.tight_layout(); plt.show()
```

The elbow is at k = 2 or 3. Setosa forms its own cluster. Versicolor and virginica overlap, so K-means splits them only partly, as the table shows.

#3.6 Support Vector Machine

Labels yᵢ ∈ {−1, +1}. A linear classifier predicts ŷ = sign(wᵀx + b)

and its decision boundary is the hyperplane wᵀx + b = 0. The distance from xᵢ to the hyperplane is |wᵀxᵢ + b| / ‖w‖

##Hard-margin SVM (separable data)

Among all separating hyperplanes, choose the one with the largest margin, meaning the largest distance to the closest point. Rescale (w, b) so that the closest points satisfy yᵢ(wᵀxᵢ + b) = 1. The margin is then 1/‖w‖ on each side (2/‖w‖ total), and maximizing it is equivalent to minimize ½‖w‖²   subject to   yᵢ(wᵀxᵢ + b) ≥ 1 for all i

This is a convex quadratic program, so it has a unique solution. Support vectors are the points with yᵢ(wᵀxᵢ + b) = 1. They lie on the margin, and they alone determine the solution.

##Example 5: Solving the hard-margin problem directly

We solve the quadratic program with scipy's constrained optimizer (SLSQP) and compare the result with scikit-learn's SVC. A very large C in SVC means essentially no margin violations are allowed, which matches the hard-margin problem.

```python
Xs, ys01 = make_blobs(n_samples=60, centers=[[-2, -1], [2, 1.5]], cluster_std=0.9, random_state=7)
ys = 2 * ys01 - 1

res = optimize.minimize(lambda v: 0.5 * v[:2] @ v[:2], x0=np.zeros(3), method="SLSQP",
                        constraints=[{"type": "ineq", "fun": lambda v: ys * (Xs @ v[:2] + v[2]) - 1}])
w_hm, b_hm = res.x[:2], res.x[2]
margins = ys * (Xs @ w_hm + b_hm)
sv_hm = np.where(np.isclose(margins, 1, atol=1e-4))[0]

svc = SVC(kernel="linear", C=1e6).fit(Xs, ys)
print("QP solution      w =", np.round(w_hm, 4), " b =", round(b_hm, 4))
print("sklearn SVC      w =", np.round(svc.coef_.ravel(), 4), " b =", round(svc.intercept_[0], 4))
print("min yᵢ(wᵀxᵢ + b) =", round(margins.min(), 6))
print("margin width 2/‖w‖ =", round(2 / np.linalg.norm(w_hm), 4))
print("support vectors (QP):     ", sv_hm)
print("support vectors (sklearn):", np.sort(svc.support_))

def plot_svm(ax, X, y, w, b, sv, title):
    ax.scatter(X[y == -1, 0], X[y == -1, 1], c="tab:blue", s=20, label="y = −1")
    ax.scatter(X[y == 1, 0], X[y == 1, 1], c="tab:red", s=20, label="y = +1")
    ax.scatter(X[sv, 0], X[sv, 1], s=140, facecolors="none", edgecolors="k", label="support vectors")
    xx = np.linspace(X[:, 0].min() - 1, X[:, 0].max() + 1, 100)
    for off, ls in [(0, "-"), (1, "--"), (-1, "--")]:
        ax.plot(xx, (off - b - w[0] * xx) / w[1], "k", ls=ls, lw=1.5 if off == 0 else 1)
    ax.set_ylim(X[:, 1].min() - 1, X[:, 1].max() + 1)
    ax.set_title(title); ax.legend(fontsize=8)

fig, ax = plt.subplots(figsize=(6.5, 5))
plot_svm(ax, Xs, ys, w_hm, b_hm, sv_hm, "Hard-margin SVM: solid = wᵀx + b = 0, dashed = ±1")
plt.show()
```

##Soft-margin SVM and the hinge loss

Real data usually can't be separated perfectly. Slack variables ξᵢ ≥ 0 allow points to fall inside the margin or on the wrong side: minimize ½‖w‖² + C ∑ξᵢ   subject to   yᵢ(wᵀxᵢ + b) ≥ 1 − ξᵢ,  ξᵢ ≥ 0

At the optimum ξᵢ = max(0, 1 − yᵢ(wᵀxᵢ + b)), so the problem becomes an unconstrained convex minimization: J(w, b) = ½‖w‖² + C ∑ max(0, 1 − yᵢ(wᵀxᵢ + b))

The second term is the hinge loss. C sets the trade-off: a large C penalizes violations heavily (narrow margin, close to hard-margin), and a small C allows more violations in exchange for a wider margin.

The hinge loss has a corner at 1, so J is convex but not differentiable there. Instead of the gradient we use a subgradient. Write M = {i : yᵢ(wᵀxᵢ + b) < 1} for the points that violate the margin. Then g_w = w − C ∑_{i ∈ M} yᵢxᵢ, g_b = −C ∑_{i ∈ M} yᵢ

Subgradient descent takes steps (w, b) ← (w, b) − αₖ(g_w, g_b) with shrinking step sizes αₖ. J is not guaranteed to decrease every step, so we keep track of the best point found.

```python
Xo, yo01 = make_blobs(n_samples=200, centers=[[-1.5, -1], [1.5, 1]], cluster_std=1.3, random_state=3)
yo = 2 * yo01 - 1
C = 1.0

def J(w, b, X, y, C):
    return 0.5 * w @ w + C * np.maximum(0, 1 - y * (X @ w + b)).sum()

w, b = np.zeros(2), 0.0
best_w, best_b, best_J = w, b, J(w, b, Xo, yo, C)
J_hist = []
for k in range(1, 20001):
    M = yo * (Xo @ w + b) < 1
    g_w = w - C * (yo[M, None] * Xo[M]).sum(axis=0)
    g_b = -C * yo[M].sum()
    alpha = 0.01 / np.sqrt(k)
    w, b = w - alpha * g_w, b - alpha * g_b
    Jk = J(w, b, Xo, yo, C)
    if Jk < best_J:
        best_w, best_b, best_J = w.copy(), b, Jk
    J_hist.append(best_J)

svc_soft = SVC(kernel="linear", C=C).fit(Xo, yo)
w_sk, b_sk = svc_soft.coef_.ravel(), svc_soft.intercept_[0]
print("subgradient   w =", np.round(best_w, 4), " b =", round(best_b, 4), " J =", round(best_J, 4))
print("sklearn SVC   w =", np.round(w_sk, 4), " b =", round(b_sk, 4), " J =", round(J(w_sk, b_sk, Xo, yo, C), 4))

fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
ax[0].semilogy(np.array(J_hist) - J(w_sk, b_sk, Xo, yo, C) + 1e-6)
ax[0].set_xlabel("iteration"); ax[0].set_ylabel("best J − J*"); ax[0].set_title("Subgradient descent on the hinge-loss objective")
plot_svm(ax[1], Xo, yo, w_sk, b_sk, svc_soft.support_, f"Soft-margin SVM, C = {C}")
plt.tight_layout(); plt.show()
```

##Example 6: Effect of C

Same data, with C ranging from 0.01 to 100. As C grows, the margin narrows and fewer points become support vectors.

```python
Cs = [0.01, 0.1, 1, 100]
fig, ax = plt.subplots(1, 4, figsize=(19, 4))
for a, c in zip(ax, Cs):
    m = SVC(kernel="linear", C=c).fit(Xo, yo)
    wc, bc = m.coef_.ravel(), m.intercept_[0]
    print(f"C = {c:<6} margin width 2/‖w‖ = {2/np.linalg.norm(wc):6.3f}   support vectors = {len(m.support_):3d}   training accuracy = {m.score(Xo, yo):.3f}")
    plot_svm(a, Xo, yo, wc, bc, m.support_, f"C = {c}")
plt.tight_layout(); plt.show()
```

##The dual problem and KKT conditions

With Lagrange multipliers αᵢ ≥ 0, the soft-margin problem has the dual maximize ∑αᵢ − ½ ∑ᵢ ∑ⱼ αᵢαⱼ yᵢyⱼ xᵢᵀxⱼ   subject to   0 ≤ αᵢ ≤ C,  ∑αᵢyᵢ = 0

The KKT conditions connect the primal and dual solutions:
- w = ∑ αᵢ yᵢ xᵢ, so w is a combination of the training points.
- αᵢ = 0 for points strictly outside the margin. Only support vectors have αᵢ > 0.
- 0 < αᵢ < C means the point lies exactly on the margin: yᵢ(wᵀxᵢ + b) = 1.
- αᵢ = C means the point is inside the margin or misclassified.

scikit-learn solves this dual problem. Its `dual_coef_` stores αᵢyᵢ for the support vectors.

```python
alpha_y = svc_soft.dual_coef_.ravel()
sv = svc_soft.support_
alpha = np.abs(alpha_y)
w_dual = alpha_y @ Xo[sv]
marg_sv = yo[sv] * (Xo[sv] @ w_sk + b_sk)

print("w from dual, ∑αᵢyᵢxᵢ =", np.round(w_dual, 6), "   w from SVC:", np.round(w_sk, 6))
print("∑αᵢyᵢ =", round(alpha_y.sum(), 10))
print("all 0 ≤ αᵢ ≤ C:", np.all((alpha >= -1e-10) & (alpha <= C + 1e-10)))
on_margin = alpha < C - 1e-6
print(f"0 < αᵢ < C: {on_margin.sum()} points, max |yᵢ(wᵀxᵢ+b) − 1| = {np.abs(marg_sv[on_margin] - 1).max():.4f}")
print(f"αᵢ = C:     {(~on_margin).sum()} points, all with yᵢ(wᵀxᵢ+b) ≤ 1: {np.all(marg_sv[~on_margin] <= 1 + 1e-6)}")

plt.scatter(marg_sv, alpha, s=20)
plt.axvline(1, color="gray", ls=":"); plt.axhline(C, color="red", ls="--", label="α = C")
plt.xlabel("yᵢ(wᵀxᵢ + b)"); plt.ylabel("αᵢ")
plt.title("Support vectors: αᵢ < C only on the margin (= 1)")
plt.legend(); plt.show()
```

##Example 7: Kernels for nonlinear boundaries

The dual depends on the data only through inner products xᵢᵀxⱼ. Replacing them with a kernel K(xᵢ, xⱼ) = φ(xᵢ)ᵀφ(xⱼ) fits a linear SVM in the feature space φ(x) without computing φ explicitly. The prediction becomes ŷ = sign( ∑ αᵢ yᵢ K(xᵢ, x) + b )

Data: two concentric circles, which no line can separate.
- Explicit feature map φ(x) = (x₁, x₂, x₁² + x₂²): the classes become linearly separable in 3-D.
- RBF kernel K(x, z) = exp(−γ‖x − z‖²): this corresponds to an infinite-dimensional φ, and the boundary becomes a curve in the original 2-D space.

```python
Xc, yc01 = make_circles(n_samples=300, factor=0.4, noise=0.08, random_state=0)
yc = 2 * yc01 - 1

lin = SVC(kernel="linear", C=1).fit(Xc, yc)
Phi = np.column_stack([Xc, (Xc ** 2).sum(axis=1)])
lin_phi = SVC(kernel="linear", C=1).fit(Phi, yc)
rbf = SVC(kernel="rbf", C=1, gamma=1).fit(Xc, yc)
print(f"linear kernel training accuracy:            {lin.score(Xc, yc):.3f}")
print(f"linear SVM on φ(x) training accuracy:       {lin_phi.score(Phi, yc):.3f}")
print(f"RBF kernel training accuracy:               {rbf.score(Xc, yc):.3f}")

g = np.linspace(-1.5, 1.5, 300)
G1, G2 = np.meshgrid(g, g)
grid = np.c_[G1.ravel(), G2.ravel()]
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
for a, model, t in [(ax[0], lin, "Linear kernel"), (ax[1], rbf, "RBF kernel")]:
    Z = model.decision_function(grid).reshape(G1.shape)
    a.contourf(G1, G2, Z, levels=20, cmap="RdBu_r", alpha=0.4)
    a.contour(G1, G2, Z, levels=[-1, 0, 1], colors="k", linestyles=["--", "-", "--"])
    a.scatter(Xc[:, 0], Xc[:, 1], c=yc, cmap="bwr", s=12)
    a.set_title(f"{t}: accuracy {model.score(Xc, yc):.2f}")
plt.tight_layout(); plt.show()
```

##Example 8: Real data (breast cancer)

569 tumors, 30 features. Features are standardized, since the margin depends on distances and therefore on the scale of each feature. Train on 75% of the data, test on the other 25%, and compare linear and RBF kernels over several values of C.

```python
bc = load_breast_cancer()
Xtr, Xte, ytr, yte = train_test_split(bc.data, bc.target, test_size=0.25, random_state=0, stratify=bc.target)
scaler = StandardScaler().fit(Xtr)
Xtr, Xte = scaler.transform(Xtr), scaler.transform(Xte)

Cs = [0.01, 0.1, 1, 10, 100]
results = {}
for kern in ["linear", "rbf"]:
    results[kern] = [(SVC(kernel=kern, C=c).fit(Xtr, ytr)) for c in Cs]
    for c, m in zip(Cs, results[kern]):
        print(f"{kern:6s} C = {c:<6} test accuracy = {m.score(Xte, yte):.3f}   support vectors = {len(m.support_)}")

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
for kern in ["linear", "rbf"]:
    ax[0].semilogx(Cs, [m.score(Xte, yte) for m in results[kern]], "o-", label=kern)
    ax[1].semilogx(Cs, [len(m.support_) for m in results[kern]], "o-", label=kern)
ax[0].set_xlabel("C"); ax[0].set_title("Test accuracy"); ax[0].legend()
ax[1].set_xlabel("C"); ax[1].set_title("Number of support vectors"); ax[1].legend()
plt.tight_layout(); plt.show()
```

K-means
- Minimizes G = ∑ ‖xᵢ − μ_c(i)‖² by repeating two steps: assign each point to its nearest center, then move each center to its cluster's mean.
- G never increases, but the result is only a local minimum and depends on the starting centers (k-means++ helps). Choose k with an elbow plot.

SVM
- Finds the hyperplane with the largest margin 2/‖w‖; only the support vectors determine it.
- Soft margin: minimize ½‖w‖² + C ∑ hinge loss. A larger C gives a narrower margin and fewer violations.
- Kernels give nonlinear boundaries.
