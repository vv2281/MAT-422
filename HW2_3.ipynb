##2.3.1 Joint Probability Distributions

A joint probability distribution describes the probabilities of two random variables occurring together.

For discrete random variables X and Y the joint probability mass function is
P(X = x, Y = y).

A valid joint probability distribution has nonnegative probabilities and all of the probabilities have to add up to 1.

The marginal distribution of X is found by summing over all possible values
of Y: P(X = x) = Σ P(X = x, Y = y)

Also, the marginal distribution of Y is P(Y = y) = Σ P(X = x, Y = y).

###Example

Suppose X is the number of heads and Y is the number of tails in two fair coin flips.

The possible outcomes are: 0 heads, 2 tails, 1 head, 1 tail, 2 heads, 0 tails.

Their probabilities are 1/4, 1/2, and 1/4.

```python
#joint probability distribution for:
#X = number of heads
#Y = number of tails

x_values = [0, 1, 2]
y_values = [0, 1, 2]

joint = np.array([
    [0,   0,   1/4],
    [0,   1/2, 0  ],
    [1/4, 0,   0  ]
])

joint_table = pd.DataFrame(
    joint,
    index=[f"X={x}" for x in x_values],
    columns=[f"Y={y}" for y in y_values]
)

print("Joint Probability Distribution:")
display(joint_table)

print("Sum of all joint probabilities:", joint.sum())
```

###Marginal Distributions

The marginal distribution of X is obtained by adding the probabilities across each row. The marginal distribution of Y is obtained by adding the probabilities
down each column.

The marginal probabilities should add up to 1.

```python
#marginal distribution of X
px = joint.sum(axis=1)

#marginal distribution of Y
py = joint.sum(axis=0)

print("Marginal Distribution P(X=x):")

for x, p in zip(x_values, px):
    print(f"P(X={x}) = {p:.2f}")

print("\nMarginal Distribution P(Y=y):")

for y, p in zip(y_values, py):
    print(f"P(Y={y}) = {p:.2f}")

print("\nCheck:")
print("Sum P(X=x) =", px.sum())
print("Sum P(Y=y) =", py.sum())
```

###Expected Values

The expected value of a function of two random variables can be calculated using the joint probability distribution.

For example: E[X] = ΣΣ x P(X=x,Y=y), E[Y] = ΣΣ y P(X=x,Y=y), E[XY] = ΣΣ xy P(X=x,Y=y)

These values help describe the behavior of the random variables.

```python
#calculate E[X], E[Y], and E[XY]

EX = sum(
    x * joint[i, j]
    for i, x in enumerate(x_values)
    for j in range(len(y_values))
)

EY = sum(
    y * joint[i, j]
    for i in range(len(x_values))
    for j, y in enumerate(y_values)
)

EXY = sum(
    x * y * joint[i, j]
    for i, x in enumerate(x_values)
    for j, y in enumerate(y_values)
)

print(f"E[X] = {EX:.2f}")
print(f"E[Y] = {EY:.2f}")
print(f"E[XY] = {EXY:.2f}")
```

###Conditional Probability

Conditional probability describes the probability of one event given that another event has occurred.

For random variables: P(X=x | Y=y) = P(X=x,Y=y) / P(Y=y)

For this example, we can calculate the distribution of X given that Y=1.

```python
#calculate P(X=x | Y=1)

y_index = 1

conditional_x_given_y1 = joint[:, y_index] / py[y_index]

print("Conditional Distribution P(X=x | Y=1):")

for x, p in zip(x_values, conditional_x_given_y1):
    print(f"P(X={x} | Y=1) = {p:.2f}")

print(
    "\nSum of conditional probabilities:",
    conditional_x_given_y1.sum()
)
```

###2.3.2 Correlation and Dependence

Two random variables are independent when their joint probability can be written as the product of their marginal probabilities: P(X=x,Y=y) = P(X=x)P(Y=y) for every possible pair (x,y).

Correlation measures the strength and direction of a linear relationship between two numerical variables.

The correlation coefficient is ρ(X,Y) = Cov(X,Y) / (σX σY).

For sample data, Pearson's correlation coefficient r is commonly used.

A correlation close to 1 indicates a strong positive linear relationship. A correlation close to -1 indicates a strong negative linear relationship. A correlation close to 0 indicates little or no linear relationship.

However, zero correlation does not necessarily mean that two variables are independent.

```python
#example of a strong positive correlation

x = np.arange(1, 11)

y = 2 * x + np.array([
    0, 1, -1, 0, 1,
    -1, 0, 1, -1, 0
])

correlation = np.corrcoef(x, y)[0, 1]

print("X:", x)
print("Y:", y)

print(f"pearson correlation r = {correlation:.4f}")

plt.figure(figsize=(7, 5))
plt.scatter(x, y)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("positive correlation")

plt.grid(True)
plt.show()
```

###Dependence Without Strong Linear Correlation

Correlation only measures linear relationships.

Consider Y = X². Here, Y is completely determined by X, so X and Y are dependent.

However, because the relationship is curved rather than linear, the correlation can be close to zero.

This demonstrates that dependent variables do not necessarily have a strong correlation.

```python
#dependent variables with a nonlinear relationship

x = np.arange(-10, 11)

y = x**2

correlation = np.corrcoef(x, y)[0, 1]

print(f"pearson correlation r = {correlation:.4f}")

plt.figure(figsize=(7, 5))

plt.scatter(x, y)

plt.xlabel("X")
plt.ylabel("Y = X²")
plt.title("dependent variables with little linear correlation")

plt.grid(True)
plt.show()
```

###Covariance and Correlation

Covariance measures how two variables change together: Cov(X,Y) = E[(X-E[X])(Y-E[Y])]

Correlation standardizes covariance using the standard deviations of the two variables. The following example generates simulated data with a positive linear relationship and calculates both covariance and correlation.

```python
#generate simulated data

rng = np.random.default_rng(42)

x = rng.normal(0, 1, 200)

y = 0.8 * x + rng.normal(0, 0.6, 200)

#calculate covariance
covariance = np.cov(x, y, ddof=1)[0, 1]

#calculate correlation
correlation = np.corrcoef(x, y)[0, 1]

print(f"Sample covariance = {covariance:.4f}")
print(f"Sample correlation = {correlation:.4f}")

plt.figure(figsize=(7, 5))

plt.scatter(x, y, alpha=0.7)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Positive Linear Association")

plt.grid(True)
plt.show()
```

###2.3.3 Random Samples

A random sample consists of observations selected from a population using a random sampling process.

A common mathematical model is an independent and identically distributed (i.i.d.) random sample: X₁, X₂, ..., Xₙ ~ F.

This means that each observation comes from the same probability distribution and the observations are independent.

For a random sample X₁, ..., Xₙ, the sample mean is X̄ = (1/n) Σ Xᵢ.

The sample variance is S^2 = (1/(n-1)) Σ(Xᵢ - X̄)^2.

The following example generates a random sample from a normal population and calculates the sample statistics.

```python
#population parameters

mu = 70
sigma = 10

#sample size
n = 30

#generate a random sample from N(70, 10^2)
sample = rng.normal(
    loc=mu,
    scale=sigma,
    size=n
)

#calculate sample statistics
sample_mean = np.mean(sample)

sample_variance = np.var(
    sample,
    ddof=1
)

sample_std = np.std(
    sample,
    ddof=1
)

print("Random Sample:")
print(np.round(sample, 2))

print(f"\nSample size n = {n}")
print(f"Sample mean = {sample_mean:.2f}")
print(f"Sample variance = {sample_variance:.2f}")
print(f"Sample standard deviation = {sample_std:.2f}")
```

###Sampling Distribution of the Sample Mean

The sample mean changes from one random sample to another.

If the population has mean μ and variance σ², then for an i.i.d. random sample of size n: E[X̄] = μ and Var(X̄) = σ²/n.

Therefore, increasing the sample size decreases the variability of the sample mean.

The next example simulates many random samples and examines their sample means.

```python
#population parameters

population_mu = 70
population_sigma = 10

#number of observations in each sample
sample_size = 30

#number of random samples
number_of_samples = 5000

#generate 5000 random samples
samples = rng.normal(
    loc=population_mu,
    scale=population_sigma,
    size=(number_of_samples, sample_size)
)

#calculate the mean of each sample
sample_means = samples.mean(axis=1)

#theoretical values
theoretical_mean = population_mu

theoretical_variance = (
    population_sigma**2 / sample_size
)

theoretical_std = (
    population_sigma / np.sqrt(sample_size)
)

print(
    f"Theoretical mean of X̄ = "
    f"{theoretical_mean:.2f}"
)

print(
    f"Theoretical variance of X̄ = "
    f"{theoretical_variance:.4f}"
)

print(
    f"Theoretical standard deviation of X̄ = "
    f"{theoretical_std:.4f}"
)

print(
    f"\nSimulated mean of X̄ = "
    f"{sample_means.mean():.4f}"
)

print(
    f"Simulated variance of X̄ = "
    f"{sample_means.var(ddof=1):.4f}"
)

print(
    f"Simulated standard deviation of X̄ = "
    f"{sample_means.std(ddof=1):.4f}"
)
```

###Visualizing the Sampling Distribution

The histogram below shows the distribution of the 5,000 simulated sample means.

The sample means are centered near the population mean of 70.

The spread of the sample means is much smaller than the spread of the individual observations because SD(X̄) = σ / sqrt(n)

```python
plt.figure(figsize=(8, 5))

plt.hist(
    sample_means,
    bins=35,
    edgecolor="black"
)

plt.axvline(
    population_mu,
    linestyle="--",
    linewidth=2,
    label="Population mean μ"
)

plt.xlabel("Sample Mean X̄")
plt.ylabel("Frequency")
plt.title("Sampling Distribution of the Sample Mean")

plt.legend()
plt.grid(True)

plt.show()
```

###Effect of Sample Size

The variance of the sample mean is Var(X̄) = σ²/n.

Therefore, as the sample size increases, the variability of the sample mean decreases.

The following simulation compares sample means for sample sizes of 5, 30, and 100.

```python
#compare different sample sizes

sample_sizes = [5, 30, 100]

results = []

for n in sample_sizes:

    simulated_samples = rng.normal(
        loc=population_mu,
        scale=population_sigma,
        size=(5000, n)
    )

    means = simulated_samples.mean(axis=1)

    results.append({
        "Sample Size": n,
        "Simulated Mean": means.mean(),
        "Simulated SD": means.std(ddof=1),
        "Theoretical SD": population_sigma / np.sqrt(n)
    })

results_df = pd.DataFrame(results)

display(results_df.round(4))
```
