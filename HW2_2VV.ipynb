##2.2.1 Probability Axioms

Probability is a numerical measure between 0 and 1 that describes how likely an event is to occur.

The three basic probability axioms include:

The probability of an event is between 0 and 1:
   
   0 ≤ P(A) ≤ 1

The probability of the entire sample space is 1:
   
   P(S) = 1

and If two events are mutually exclusive, then:
   
   P(A ∪ B) = P(A) + P(B)

We can demonstrate these axioms using a six sided die.

```python
import numpy as np
import matplotlib.pyplot as plt

#define the possible outcomes when rolling a fair six-sided die
sample_space = np.array([1, 2, 3, 4, 5, 6])

#every outcome has the same probability
probability_each = 1 / len(sample_space)

print("Sample space:", sample_space)
print("Probability of each outcome:", probability_each)
print("Probability of the entire sample space:",
      probability_each * len(sample_space))
```

```python
#axiom 1:
#the probability of an event must be between 0 and 1.
event_A = np.array([2, 4, 6])  #rolling an even number

P_A = len(event_A) / len(sample_space)

print("Event A: rolling an even number")
print("P(A) =", P_A)

#check that tjhe probability is between 0 and 1
print("Is 0 <= P(A) <= 1?", 0 <= P_A <= 1)
```

```python
#axiom 2:
#the probability of the entire sample space is equal to 1.

P_sample_space = len(sample_space) / len(sample_space)

print("P(S) =", P_sample_space)
print("The probability of the entire sample space is 1.")
```

```python
#axiom 3:
#for mutually exclusive events: P(A union B) = P(A) + P(B)

#event A: rolling a 1
A = {1}

#event B: rolling a 6
B = {6}

#A and B cannot happen at the same time, so they are mutually exclusive.

P_A = len(A) / len(sample_space)
P_B = len(B) / len(sample_space)

P_A_union_B = P_A + P_B

print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A union B) =", P_A_union_B)
```

##2.2.2 Conditional Probability

Conditional probability measures the probability of an event occurring given that another event has already occurred.

The formula is:

P(A | B) = P(A ∩ B) / P(B)

For instance, imagine that we roll a six-sided die.

And so let: A = rolling an even number and B = rolling a number greater than 3

We can calculate the probability of A given B.

```python
#define events

#A = even number
A = {2, 4, 6}

#and B = number greater than 3
B = {4, 5, 6}

#intersection of A and B
A_intersection_B = A.intersection(B)

#calculate probabilities
P_B = len(B) / len(sample_space)
P_A_intersection_B = len(A_intersection_B) / len(sample_space)

#conditional probability formula
P_A_given_B = P_A_intersection_B / P_B

print("Event A (even numbers):", A)
print("Event B (numbers greater than 3):", B)
print("A intersection B:", A_intersection_B)

print("\nP(B) =", P_B)
print("P(A intersection B) =", P_A_intersection_B)
print("P(A | B) =", P_A_given_B)
```

The result shows that: P(A | B) = 2/3

Once we know that the number is greater than 3, the possible
outcomes are only 4, 5, and 6. Two of these three outcomes
are even.

```python
#visualize the outcomes involved in the conditional probability

outcomes = [1, 2, 3, 4, 5, 6]
probabilities = [1/6] * 6

plt.figure(figsize=(8, 5))
plt.bar(outcomes, probabilities)

plt.xlabel("Die Outcome")
plt.ylabel("Probability")
plt.title("Probability of Each Die Outcome")
plt.xticks(outcomes)

plt.show()
```

##2.2.3 Discrete Random Variables

A discrete random variable can take a countable number of
possible values. For example, when rolling a fair six-sided die, the random
variable X can take the values:

X = 1, 2, 3, 4, 5, 6

The probability mass function (PMF) gives the probability
associated with each possible value.

For a fair die: P(X = x) = 1/6

```python
#define the possible values of the discrete random variable X
X = np.array([1, 2, 3, 4, 5, 6])

#since the die is fair, every value has probability 1/6
P_X = np.array([1/6] * 6)

print("Discrete Random Variable X")
print("--------------------------")

for value, probability in zip(X, P_X):
    print(f"P(X = {value}) = {probability:.4f}")

#check that all probabilities add to 1
print("\nSum of probabilities =", P_X.sum())
```

#Expected Value

The expected value of a discrete random variable is: E[X] = Σ x P(X = x)

It represents the long-run average value of the random variable.

```python
#calculate the expected value
expected_value = np.sum(X * P_X)

print("Expected value E[X] =", expected_value)
```

For a fair six-sided die, the expected value is 3.5.

However, this does not mean that a single roll will produce 3.5. Instead, it means that if the die is rolled many times, the average of the results will approach 3.5.

```python
#calculate the variance and standard deviation

#variance formula: Var(X) = E[(X - E[X])^2]

variance = np.sum((X - expected_value)**2 * P_X)

#standard deviation is the square root of variance
standard_deviation = np.sqrt(variance)

print("Variance =", variance)
print("Standard deviation =", standard_deviation)

#plot the probability mass function

plt.figure(figsize=(8, 5))
plt.bar(X, P_X)

plt.xlabel("X")
plt.ylabel("P(X = x)")
plt.title("Probability Mass Function of a Fair Die")
plt.xticks(X)

plt.show()
```

##2.2.4 Continuous Random Variables

A continuous random variable can take any value within an interval.

Examples include: Height, weight, temperature, time, and distance

Unlike a discrete random variable, a continuous random variable has infinitely many possible values.

For a continuous random variable, probabilities are represented by the area under a probability density function (PDF).

For this example, we will use a standard normal distribution.

```python
#import the normal distribution from SciPy
from scipy.stats import norm

#create values along the x-axis
x = np.linspace(-4, 4, 1000)

#calculate the probability density for each x value
pdf = norm.pdf(x)

#plotting the normal probability density function
plt.figure(figsize=(9, 5))
plt.plot(x, pdf)

plt.xlabel("X")
plt.ylabel("Probability Density")
plt.title("Standard Normal Probability Density Function")

plt.show()
```

###Probability for a Continuous Random Variable

For a continuous random variable, the probability that X falls between two values a and b can be calculated using the CDF: P(a ≤ X ≤ b) = F(b) - F(a)

For the standard normal distribution, we can calculate the probability that X is between -1 and 1.

```python
#calculate P(-1 <= X <= 1)

lower = -1
upper = 1

#norm.cdf() gives the cumulative probability up to a value
probability = norm.cdf(upper) - norm.cdf(lower)

print("P(-1 <= X <= 1) =", probability)

#visualize the probability between -1 and 1

x = np.linspace(-4, 4, 1000)
pdf = norm.pdf(x)

plt.figure(figsize=(9, 5))

#plot the normal distribution
plt.plot(x, pdf)

#shade the area between -1 and 1
x_fill = np.linspace(-1, 1, 500)
y_fill = norm.pdf(x_fill)

plt.fill_between(x_fill, y_fill, alpha=0.5)

plt.xlabel("X")
plt.ylabel("Probability Density")
plt.title("P(-1 ≤ X ≤ 1)")

plt.show()
```
