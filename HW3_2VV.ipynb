#Limits and Continuity, Derivatives, and Taylor’s Theorem


```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

##3.2.1 Limits and Continuity

A limit describes the value a function approaches as the input gets closer to a particular point, even if the function is not actually defined at that point.

lim(x→a) f(x) = L means that f(x) gets arbitrarily close to L as x gets close to a. More formally, for every ε > 0 there exists a δ > 0 such that 0 < |x - a| < δ implies |f(x) - L| < ε.

A function f is continuous at a point a when lim(x→a) f(x) = f(a), meaning the limit exists, f(a) exists, and the two values agree. A function can fail to be continuous in three common ways: a removable discontinuity, a jump discontinuity, or an essential (infinite) discontinuity.

##Example

Consider f(x) = sin(x)/x. This function is undefined at x = 0 because of division by zero. We can check whether the limit exists by evaluating f(x) at values very close to 0 from both sides.

```python
#f(x) = sin(x)/x is undefined at x=0
#check values approaching 0 from both sides

def f_sinc(x):
    return np.sin(x) / x

h = np.array([1, 0.5, 0.1, 0.01, 0.001, 0.0001])

print("x approaching 0 from the right:")
for hi in h:
    print(f"f({hi}) = {f_sinc(hi):.6f}")

print("\nx approaching 0 from the left:")
for hi in h:
    print(f"f({-hi}) = {f_sinc(-hi):.6f}")

print("\nBoth sides approach 1, so the limit as x approaches 0 is 1, even though f(0) is undefined.")

x_plot = np.linspace(-10, 10, 2000)
x_plot = x_plot[np.abs(x_plot) > 1e-6]

plt.figure(figsize=(7, 5))
plt.plot(x_plot, f_sinc(x_plot))
plt.scatter([0], [1], facecolors="none", edgecolors="red", s=80, label="limit = 1 (f(0) undefined)")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = sin(x)/x: removable discontinuity at x = 0")

plt.legend()
plt.grid(True)
plt.show()
```

###Jump Discontinuity

Not every limit exists. For a jump discontinuity, the left-hand and right-hand limits both exist but do not agree, so the overall limit does not exist. The sign function is a simple example.

```python
#sign function: left-hand and right-hand limits disagree

def f_sign(x):
    return np.where(x > 0, 1.0, np.where(x < 0, -1.0, 0.0))

h = np.array([1, 0.1, 0.01, 0.001])

print("right-hand limit samples, f(x) as x -> 0+:", f_sign(h))
print("left-hand limit samples, f(x) as x -> 0-:", f_sign(-h))

print("\nThe right-hand limit is 1 and the left-hand limit is -1. Since they disagree, the limit as x approaches 0 does not exist.")

x_plot = np.linspace(-3, 3, 400)

plt.figure(figsize=(7, 5))
plt.plot(x_plot[x_plot < 0], f_sign(x_plot[x_plot < 0]))
plt.plot(x_plot[x_plot > 0], f_sign(x_plot[x_plot > 0]))
plt.scatter([0, 0], [-1, 1], facecolors="none", edgecolors="red", s=80)
plt.scatter([0], [0], color="red")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("sign(x): jump discontinuity at x = 0")

plt.grid(True)
plt.show()
```

###Essential Discontinuity

An essential, or infinite, discontinuity occurs when the one-sided limits do not approach a finite value at all. The function f(x) = 1/x behaves this way at x = 0.

```python
#f(x) = 1/x diverges on both sides of 0

h = np.array([1, 0.1, 0.01, 0.001, 0.0001])

print("x -> 0+:", 1 / h)
print("x -> 0-:", 1 / (-h))

print("\nNeither one-sided limit approaches a finite value, so the limit as x approaches 0 does not exist.")

x_pos = np.linspace(0.05, 3, 300)
x_neg = np.linspace(-3, -0.05, 300)

plt.figure(figsize=(7, 5))
plt.plot(x_pos, 1 / x_pos)
plt.plot(x_neg, 1 / x_neg)
plt.axvline(0, linestyle="--", linewidth=1, label="vertical asymptote x = 0")

plt.ylim(-20, 20)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = 1/x: essential discontinuity at x = 0")

plt.legend()
plt.grid(True)
plt.show()
```

###Checking Continuity with Epsilon and Delta

The epsilon-delta definition can be checked directly for a continuous function. For f(x) = x^2 at a = 2, f(a) = 4. Choosing epsilon = 0.1, we can find a delta small enough that every x within delta of 2 keeps f(x) within epsilon of 4.

```python
#|x^2 - 4| = |x-2||x+2|
#for |x-2| < 1, |x+2| < 5, so delta = epsilon/5 works

a, fa = 2.0, 4.0
epsilon = 0.1
delta = epsilon / 5

x_band = np.linspace(a - delta, a + delta, 9)
max_diff = np.max(np.abs(x_band**2 - fa))

print(f"epsilon = {epsilon}, delta = {delta:.4f}")
print(f"max |f(x) - f(a)| for x within delta of a = {max_diff:.4f}")
print(f"condition holds (max diff < epsilon): {max_diff < epsilon}")

x_grid = np.linspace(1.5, 2.5, 300)

plt.figure(figsize=(7, 5))
plt.plot(x_grid, x_grid**2, color="black", label="f(x) = x^2")
plt.axhspan(fa - epsilon, fa + epsilon, alpha=0.15, label="epsilon band around f(a)")
plt.axvspan(a - delta, a + delta, alpha=0.15, label="delta band around a")
plt.scatter([a], [fa], color="black")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("delta band around a maps inside epsilon band around f(a)")

plt.legend(fontsize=8)
plt.grid(True)
plt.show()
```

##3.2.2 Derivatives

The derivative of f at a point a is defined as a limit:

f'(a) = lim(h→0) [f(a+h) - f(a)] / h

This limit, when it exists, gives the slope of the tangent line to f at a. In practice we can approximate it using a central difference, [f(a+h) - f(a-h)] / (2h), which converges to f'(a) faster than a one-sided difference.

Differentiability is a stronger condition than continuity. A function can be continuous at a point without being differentiable there.

###Example

Consider f(x) = x^3 at a = 1. The exact derivative is 3a^2 = 3. We can approximate it with central differences using smaller and smaller values of h and see the approximation converge to the exact slope.

```python
#central difference approximation of the derivative

def central_diff(f, a, h):
    return (f(a + h) - f(a - h)) / (2 * h)

def f_cubic(x):
    return x**3

a = 1.0
exact = 3 * a**2

h_vals = np.array([1, 0.5, 0.1, 0.01, 0.001, 0.0001])

for h in h_vals:
    slope = central_diff(f_cubic, a, h)
    print(f"h = {h}: central-difference slope = {slope:.8f} (exact = {exact})")

x_range = np.linspace(0, 2, 200)

plt.figure(figsize=(7, 5))
plt.plot(x_range, f_cubic(x_range), color="black", label="f(x) = x^3")
plt.plot(x_range, f_cubic(a) + exact * (x_range - a), "--", color="red", label=f"tangent, slope = {exact}")

for h, color in zip([1, 0.5, 0.1], ["gold", "orange", "steelblue"]):
    slope = central_diff(f_cubic, a, h)
    plt.plot(x_range, f_cubic(a) + slope * (x_range - a), color=color, alpha=0.8, label=f"secant h={h}, slope={slope:.3f}")

plt.scatter([a], [f_cubic(a)], color="black")
plt.ylim(-2, 10)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("secant lines converging to the tangent line as h decreases")

plt.legend(fontsize=8)
plt.grid(True)
plt.show()
```

###Differentiability vs. Continuity

The function f(x) = |x| is continuous at x = 0, but it is not differentiable there. Approaching from the right gives a different slope than approaching from the left.

```python
#forward and backward difference quotients for f(x) = |x| at x = 0

def f_abs(x):
    return np.abs(x)

h_vals = np.array([1, 0.1, 0.01, 0.001])

forward = (f_abs(h_vals) - f_abs(0)) / h_vals
backward = (f_abs(0) - f_abs(-h_vals)) / h_vals

print("forward difference quotient (h -> 0+):", forward)
print("backward difference quotient (h -> 0-):", backward)

print("\nThe forward quotient approaches 1 and the backward quotient approaches -1.")
print("Since they disagree, f(x) = |x| is not differentiable at x = 0, even though it is continuous there.")
```

###Checking the Order of Accuracy

The central difference formula has error that shrinks proportionally to h^2 as h gets smaller. We can check this by comparing the approximation error to the exact derivative for a few different functions.

```python
#central-difference error should shrink like h^2

test_cases = {
    "sin(x), a=1": (np.sin, np.cos, 1.0),
    "exp(x), a=1": (np.exp, np.exp, 1.0),
    "x^3, a=2": (lambda x: x**3, lambda x: 3 * x**2, 2.0),
}

h_grid = np.logspace(-1, -6, 12)

plt.figure(figsize=(7, 5))

for label, (f, fprime, a) in test_cases.items():
    exact = fprime(a)
    errors = np.abs([central_diff(f, a, h) - exact for h in h_grid])
    plt.loglog(h_grid, errors, marker="o", markersize=3, label=label)

plt.loglog(h_grid, 0.1 * h_grid**2, "--", color="black", label="h^2 reference")

plt.xlabel("step size h")
plt.ylabel("absolute error")
plt.title("central-difference error shrinks proportionally to h^2")

plt.legend(fontsize=8)
plt.grid(True)
plt.show()
```

###Product Rule and Chain Rule

The product rule states (fg)' = f'g + fg', and the chain rule states that the derivative of f(g(x)) is f'(g(x))g'(x). Both can be checked numerically using central differences.

```python
#check the product rule and chain rule numerically

def f1(x):
    return np.sin(x)

def f2(x):
    return x**2 + 1

def product(x):
    return f1(x) * f2(x)

def chain(x):
    return np.sin(f2(x))

a, h = 1.3, 1e-5

f1_prime = central_diff(f1, a, h)
f2_prime = central_diff(f2, a, h)

print("product rule:")
print(f"  numerical (fg)'(a) = {central_diff(product, a, h):.6f}")
print(f"  f'(a)g(a) + f(a)g'(a) = {f1_prime * f2(a) + f1(a) * f2_prime:.6f}")

print("\nchain rule:")
print(f"  numerical d/dx sin(g(x)) = {central_diff(chain, a, h):.6f}")
print(f"  f1'(g(a)) * g'(a) = {np.cos(f2(a)) * f2_prime:.6f}")
```

###Application: Gradient Descent

Gradient descent uses the derivative to minimize a function. At each step, we move against the direction of the derivative: x(n+1) = x(n) - α f'(x(n)). This is the basic idea behind fitting many models in data science.

```python
#minimize f(x) = (x-3)^2 + 1 using gradient descent

def loss(x):
    return (x - 3)**2 + 1

def grad(x):
    return 2 * (x - 3)

x = -4.0
learning_rate = 0.2

path = [x]

for step in range(25):
    x = x - learning_rate * grad(x)
    path.append(x)

path = np.array(path)

print(f"starting x = {path[0]}")
print(f"after {len(path) - 1} steps, x = {path[-1]:.6f} (true minimum is at x = 3)")

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

x_curve = np.linspace(-5, 8, 200)
axes[0].plot(x_curve, loss(x_curve))
axes[0].plot(path, loss(path), "o-", color="red", markersize=4, linewidth=1)
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")
axes[0].set_title("gradient descent path")

axes[1].plot(loss(path), "o-", color="darkorange")
axes[1].set_xlabel("iteration")
axes[1].set_ylabel("f(x)")
axes[1].set_title("loss vs. iteration")

plt.tight_layout()
plt.show()
```

##3.2.3 Taylor's Theorem

The degree-n Taylor polynomial of f about a is

Tₙ(x) = Σ (f^(k)(a) / k!) (x-a)^k, for k = 0 to n

This polynomial approximates f near a. The error of the approximation, called the Lagrange remainder, is

Rₙ(x) = f(x) - Tₙ(x) = f^(n+1)(c) / (n+1)! * (x-a)^(n+1)

for some c between a and x.

###Example

Every derivative of e^x is e^x, so f^(k)(0) = 1 for all k. The Taylor polynomial about a = 0 is just the familiar power series for e^x. We can see how increasing the degree improves the approximation.

```python
#taylor polynomial for exp(x) about a = 0

def taylor_exp(x, n):
    return sum(x**k / math.factorial(k) for k in range(n + 1))

x_grid = np.linspace(-3, 3, 400)

plt.figure(figsize=(7, 5))
plt.plot(x_grid, np.exp(x_grid), color="black", linewidth=2, label="exp(x)")

for n, color in zip([1, 2, 4, 8], ["gold", "orange", "steelblue", "seagreen"]):
    approx = [taylor_exp(x, n) for x in x_grid]
    plt.plot(x_grid, approx, "--", color=color, label=f"degree {n}")

plt.ylim(-2, 20)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("taylor polynomials of exp(x) about a = 0")

plt.legend()
plt.grid(True)
plt.show()
```

###Taylor Series for sin(x)

The derivatives of sin(x) cycle through sin, cos, -sin, -cos, so only the odd-degree terms appear in the Taylor polynomial about a = 0, with alternating signs.

```python
#taylor polynomial for sin(x) about a = 0

def taylor_sin(x, n):
    total = 0.0
    for k in range(n + 1):
        if k % 2 == 1:
            sign = 1 if k % 4 == 1 else -1
            total += sign * x**k / math.factorial(k)
    return total

x_grid = np.linspace(-2 * np.pi, 2 * np.pi, 400)

plt.figure(figsize=(7, 5))
plt.plot(x_grid, np.sin(x_grid), color="black", linewidth=2, label="sin(x)")

for n, color in zip([1, 3, 5, 9], ["gold", "orange", "steelblue", "seagreen"]):
    approx = [taylor_sin(x, n) for x in x_grid]
    plt.plot(x_grid, approx, "--", color=color, label=f"degree {n}")

plt.ylim(-3, 3)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("taylor polynomials of sin(x) about a = 0")

plt.legend()
plt.grid(True)
plt.show()
```

###Remainder and Error

The Lagrange remainder gives a bound on how large the error can be. For exp(x) evaluated at x0 = 2 with a = 0, the (n+1)th derivative is still exp(c) for some c between 0 and 2, which is at most exp(2). This gives an upper bound we can compare against the actual error.

```python
#actual error vs the Lagrange remainder bound for exp(x) at x0 = 2

x0 = 2.0
degrees = np.arange(0, 15)

actual_error = [abs(math.exp(x0) - taylor_exp(x0, n)) for n in degrees]
bound = [math.exp(x0) * x0**(n + 1) / math.factorial(n + 1) for n in degrees]

plt.figure(figsize=(7, 5))
plt.semilogy(degrees, actual_error, "o-", label="actual error")
plt.semilogy(degrees, bound, "s--", label="Lagrange remainder bound")

plt.xlabel("degree n")
plt.ylabel("error (log scale)")
plt.title("taylor remainder at x = 2: actual error vs. theoretical bound")

plt.legend()
plt.grid(True)
plt.show()

print("actual error <= bound for every degree tested:", all(a <= b + 1e-12 for a, b in zip(actual_error, bound)))
```

###Application: Newton's Method

Newton's method minimizes a function by repeatedly jumping to the minimum of its second-order Taylor approximation at the current point. This gives the update rule x(n+1) = x(n) - f'(x(n)) / f''(x(n)).

```python
#newton's method on f(x) = x^4/4 - x^2

def f_quartic(x):
    return x**4 / 4 - x**2

def f_prime(x):
    return x**3 - 2 * x

def f_second(x):
    return 3 * x**2 - 2

x0 = 3.0
xs = [x0]

for step in range(6):
    x_next = xs[-1] - f_prime(xs[-1]) / f_second(xs[-1])
    xs.append(x_next)

print(f"{'iteration':>10} {'x':>12} {'f(x)':>12}")
for i, x in enumerate(xs):
    print(f"{i:>10} {x:>12.6f} {f_quartic(x):>12.6f}")

print(f"\ntrue local minimum is at x = sqrt(2) = {math.sqrt(2):.6f}")

T2 = lambda x: f_quartic(x0) + f_prime(x0) * (x - x0) + f_second(x0) / 2 * (x - x0)**2
x1 = x0 - f_prime(x0) / f_second(x0)

x_grid = np.linspace(0, 4, 300)

plt.figure(figsize=(7, 5))
plt.plot(x_grid, f_quartic(x_grid), color="black", linewidth=2, label="f(x) = x^4/4 - x^2")
plt.plot(x_grid, T2(x_grid), "--", color="red", label="2nd-order taylor approximation at x0 = 3")
plt.scatter([x0, x1], [f_quartic(x0), T2(x1)], color=["black", "red"])

plt.ylim(-3, 10)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("newton's method: minimizing the local quadratic approximation")

plt.legend()
plt.grid(True)
plt.show()
```
